## Review and export

24 clips / 1,358 frames. Doom particles use a deterministic port of the original XML emitter parameters and original textures; the native nuke supplies the full-screen flash. Sleep poses are exported but unused in Nacht's awake Doom-shroom gameplay. Original source image resolution is preserved through atlas packing.

The 35.4× historical Peashooter benchmark does not apply to this new workload. Asset validation does not replace in-game combat testing.


```python
import base64,zipfile
from IPython.display import display,HTML,FileLink
for filename,report in [('amd-all-plants.zip',roster),('amd-doomshroom.zip',doom)]:
    with zipfile.ZipFile(filename) as z:
        for name in report['clips']:
            data=base64.b64encode(z.read(name+'.webp')).decode()
            display(HTML('<div style="padding:8px;background:#777"><b>'+name+'</b><br><img src="data:image/webp;base64,'+data+'"></div>'))
    display(FileLink(filename))
```
